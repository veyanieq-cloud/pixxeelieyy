1. Массивы NumPy

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(42)

# Создаем 100 треков
danceability = rng.uniform(0.1, 1.0, 100)  # танцевальность (0-1)
energy = rng.uniform(0.1, 1.0, 100)        # энергичность (0-1)
duration = rng.uniform(120, 360, 100)      # длительность в секундах (2-6 мин)
popularity = rng.integers(0, 100, 100)     # популярность (0-100)

print("Средняя популярность:", popularity.mean())
print("Медианная длительность:", np.median(duration))
```

```python
# Матрица корреляций
matrix = np.corrcoef([danceability, energy, duration, popularity])
print("Матрица корреляций:\n", matrix)

# Маска для популярных треков (популярность > 50)
mask = popularity > 50
print("Количество популярных треков:", mask.sum())

# Сравнение средних
print("Средняя энергичность популярных:", energy[mask].mean())
print("Средняя энергичность остальных:", energy[~mask].mean())
```

2. Создание CSV и работа с pandas

```python
# Создаем DataFrame с "грязными" данными для задания
data = {
    "artist": ["Artist A", "Artist B", "Artist A", "Artist C", "Artist D"] * 5,
    "genre": ["Pop", "Rock", "Pop", "Jazz", "Rock"] * 5,
    "year": [2018, 2019, 2018, 2020, 2021] * 5,
    "duration_ms": [210000, 250000, None, 180000, 300000] * 5,
    "popularity": [75, 60, 80, 45, 90] * 5
}
df_tracks = pd.DataFrame(data)
# Добавим немного "грязи"
df_tracks.loc[2, "genre"] = " pop "  # лишние пробелы и регистр
df_tracks.loc[5, "duration_ms"] = -1000  # невозможное значение

# Сохраняем в файл
df_tracks.to_csv("tracks.csv", index=False)
print("Файл tracks.csv успешно создан!")
```

```python
# Загружаем файл
df = pd.read_csv("tracks.csv")

# Переводим мс в минуты
df["duration_min"] = df["duration_ms"] / 60000

# Обрабатываем пропуски (медианой)
df["duration_min"] = df["duration_min"].fillna(df["duration_min"].median())

# Чистим жанры (убираем пробелы и приводим к единому виду)
df["genre"] = df["genre"].str.strip().str.title()

# Удаляем невозможные значения (отрицательная длительность)
df = df[df["duration_min"] > 0]

# Группировка по жанрам и годам
print("Средняя популярность по жанрам:\n", df.groupby("genre")["popularity"].mean())

# Самые представленные исполнители
print("\nТоп исполнителей:\n", df["artist"].value_counts().head(3))
```

3. Графики Matplotlib

```python
# 1. Линейный график средней популярности по годам
avg_pop_year = df.groupby("year")["popularity"].mean()
plt.figure(figsize=(8, 4))
plt.plot(avg_pop_year.index, avg_pop_year.values, marker="o", color="b")
plt.title("Средняя популярность по годам")
plt.xlabel("Год")
plt.ylabel("Популярность")
plt.grid(True)
plt.show()

# 2. Столбчатая диаграмма числа треков по жанрам
genre_counts = df["genre"].value_counts()
plt.figure(figsize=(8, 4))
plt.bar(genre_counts.index, genre_counts.values, color="orange")
plt.title("Количество треков по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Количество")
plt.show()
```

```python
# 3. Гистограмма длительности
plt.figure(figsize=(8, 4))
plt.hist(df["duration_min"], bins=10, edgecolor="black", color="green")
plt.title("Распределение длительности треков")
plt.xlabel("Длительность (мин)")
plt.ylabel("Частота")
plt.show()

# 4. Диаграмма рассеяния «энергичность – популярность»
# (Так как энергичности в нашем df нет, сгенерируем её для примера)
df["energy"] = rng.uniform(0.1, 1.0, len(df))
plt.figure(figsize=(8, 4))
plt.scatter(df["energy"], df["popularity"], alpha=0.6, color="red")
plt.title("Связь энергичности и популярности")
plt.xlabel("Энергичность")
plt.ylabel("Популярность")
plt.show()
```

4. Графики seaborn

```python
# Boxplot популярности по жанрам
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="genre", y="popularity", palette="Set2")
plt.title("Популярность по жанрам (Boxplot)")
plt.show()

# Histplot танцевальности (сгенерируем для примера)
df["danceability"] = rng.uniform(0.1, 1.0, len(df))
plt.figure(figsize=(8, 4))
sns.histplot(df["danceability"], kde=True, color="purple")
plt.title("Распределение танцевальности")
plt.show()

# Scatterplot с цветом по жанру
plt.figure(figsize=(8, 4))
sns.scatterplot(data=df, x="energy", y="popularity", hue="genre", s=100)
plt.title("Энергичность vs Популярность (цвет - жанр)")
plt.show()
```

```python
# Выбираем только числовые столбцы для корреляции
numeric_df = df[["popularity", "duration_min", "energy", "danceability", "year"]]
plt.figure(figsize=(6, 5))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Матрица корреляций характеристик")
plt.show()
```

5. Самостоятельное исследование

```python
import os
import kagglehub
path = kagglehub.dataset_download("maharshipandya/-spotify-tracks-dataset")
df_kaggle = pd.read_csv(os.path.join(path, "dataset.csv"))
print("Data shape:", df_kaggle.shape)
df_kaggle.head()
df_kaggle.info()
df_kaggle.describe()
```

```python
# Таблица пропусков с долями
missing = df_kaggle.isna().sum().to_frame(name="Пропуски")
missing["Доля, %"] = (missing["Пропуски"] / len(df_kaggle) * 100).round(2)
print(missing[missing["Пропуски"] > 0])

# Удаление дубликатов
df_kaggle = df_kaggle.drop_duplicates()
```

```python
# Пример для признака 'popularity'
q1, q3 = df_kaggle["popularity"].quantile([0.25, 0.75])
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr

# Оставляем только нормальные значения
df_kaggle = df_kaggle[(df_kaggle["popularity"] >= low) & (df_kaggle["popularity"] <= high)]
```

```python
# Производный признак: длительность в минутах
df_kaggle["duration_min"] = df_kaggle["duration_ms"] / 60000

# График 1: Топ-10 исполнителей
top_artists = df_kaggle["artists"].value_counts().head(10)
plt.figure(figsize=(10, 4))
sns.barplot(x=top_artists.values, y=top_artists.index, palette="viridis")
plt.title("Топ-10 исполнителей по числу треков")
plt.show()

# График 2: Корреляция популярности и танцевальности
plt.figure(figsize=(8, 4))
sns.regplot(data=df_kaggle, x="danceability", y="popularity", scatter_kws={'alpha':0.3})
plt.title("Связь танцевальности и популярности")
plt.show()
```
